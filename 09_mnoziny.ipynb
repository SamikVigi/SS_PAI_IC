# Množiny (Sets)

#### Set (Množina) vs Dictionary (Slovník)

##### Set (množina) - iba hodnoty
* Obsahuje len hodnoty (bez kľúčov)
* Nemá duplicity
* Nemá poradie (nezoradené)

```python
pythonfarby = {'červená', 'zelená', 'modrá'}
cisla = {1, 2, 3, 4, 5}

print(type(pythonfarby))
print(type(cisla))
```

#### POZOR na prázdne zátvorky!
Prázdne `{}` Python vždy považuje za **slovník**, nie množinu!

```python
prazdny_slovnik = {}
print(type(prazdny_slovnik))

prazdna_mnozina = set() 
print(type(prazdna_mnozina))
```

### Prienik (intersection) - spoločné prvky ∩

* Metóda `intersection()` alebo operátor `&`

```python
mnozina1 = {1, 2, 3, 4, 5}
mnozina2 = {4, 5, 6, 7, 8}

prienik = mnozina1 & mnozina2
print(prienik) 
```

### Zjednotenie (union) - všetky prvky ∪

* Metóda `union()` alebo operátor `|`

```python
zjednotenie = mnozina1 | mnozina2
print(zjednotenie)
```

### Úloha 1: Doplň kód

Doplň chýbajúce časti kódu pre vytvorenie množiny farieb.

```python
# Vytvor množinu farieb 
# 'žltá', 'zelená', 'čierna', 'modrá'
farby = {'žltá', 'zelená', 'čierna', 'modrá'}
print(farby)

```

```python
farby.add('orandžova')
print(farby)
```

```python
farby.add('zelená')
print(farby)
```

```python
farby.remove('žltá')
print(farby)
```

### Úloha 2: Oprav kód

Nasledujúci kód má chyby. Nájdi ich a oprav.

```python
# Chceme nájsť spoločné prvky (prienik)
cisla1 = {1, 2, 3, 4, 5}
cisla2 = {4, 5, 6, 7, 8}

spolocne = cisla1 + cisla2
print(spolocne)
```

```python
spolocne = cisla1 | cisla2
print(spolocne)
```

### Úloha 3: Vytvor kód

Vytvor dve množiny `pj1` ('Python', 'Java', 'C++', 'JavaScript') a `pj2` ('Python', 'Ruby', 'JavaScript', 'Go').
Potom nájdi a vypíš: Spoločné programovacie jazyky, Všetky programovacie jazyky dohromady a Jazyky, ktoré sú len v prvej množine.

```python
# Napíš svoj kód tu
pj1 = {'Python', 'Java', 'C++', 'JavaScript'}
pj2 = {'Python', 'Ruby', 'JavaScript', 'Go'}
```

```python
zjednotenie = pj1 & pj2
print(zjednotenie)
```

```python
spolocne = pj1 | pj2
print(spolocne)
```
